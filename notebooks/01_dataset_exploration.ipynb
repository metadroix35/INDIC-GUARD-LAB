# 01: Dataset Exploration & Multilingual Safety Distribution

This notebook inspects schema integrity, linguistic coverage, and balance across safety categories.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.data.loader import load_or_create_dataset
from src.data.validation import DatasetValidator

# Load benchmark dataset
df = load_or_create_dataset()
print(f"Dataset shape: {df.shape}")
df.head(5)
```

```python
# Validate dataset schema and labels
validator = DatasetValidator()
report = validator.check_data_quality(df)
print("Data Quality Summary:")
for k, v in report.items():
    if k != "missing_values":
        print(f"  {k}: {v}")
```

```python
# Visualize Language and Risk Distribution
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
df['language'].value_counts().plot(kind='bar', ax=ax1, color='#0284c7', title='Language Breakdown')
df['risk_level'].value_counts().plot(kind='bar', ax=ax2, color=['#10b981', '#ef4444', '#f59e0b'], title='Risk Distribution')
plt.tight_layout()
plt.show()
```
